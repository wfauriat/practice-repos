```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import matplotlib.pyplot as plt
import seaborn as sns

from src.loaders import load_tips, by_day_summary

sns.set_theme(context='notebook', style='whitegrid')
```

## 1. Load and sanity-check

*What we want to know*: are there any obvious data quality issues before we proceed?

```python
df = load_tips()
df.head()
```

```python
df.describe(include='all').T
```

## 2. Tip percentage by day

*What we want to know*: do diners on certain days tip differently?

```python
summary = by_day_summary(df)
summary
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
sns.barplot(data=summary, x='day', y='mean_tip_pct', ax=ax)
ax.set_ylabel('Mean tip %')
ax.set_title('Tip percentage by day of week')
plt.show()
```

## 3. Party size vs total bill

*What we want to know*: is bigger party → bigger bill linear, sub-linear, or not really a thing?

```python
fig, ax = plt.subplots(figsize=(8, 4))
sns.scatterplot(data=df, x='size', y='total_bill', hue='day', ax=ax)
ax.set_title('Party size vs total bill')
plt.show()
```
